# Granger vs PEID MLP Comparison

This notebook runs the controlled comparison between lag-ablation Granger pairwise edges and PEID pairwise/synergy edges from the same learned MLP transition model.

Set `GRANGER_PEID_NOTEBOOK_SMOKE=1` to force the smallest smoke grid.

```python
from __future__ import annotations

import json
import os
import sys
from pathlib import Path

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next(
    path for path in candidates
    if (path / "scripts" / "compare_granger_peid_mlp.py").exists()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts.compare_granger_peid_mlp import run_comparison_grid
```

## Run

The default notebook path reuses the script defaults. Smoke mode keeps the run small enough for routine checks; full mode should be launched from the CLI when a longer sweep is needed.

```python
smoke = os.environ.get("GRANGER_PEID_NOTEBOOK_SMOKE", "1") != "0"
mode = "smoke" if smoke else "full"

output = run_comparison_grid(mode=mode)
output
```

## Summary Table

```python
summary_path = Path(output["summary_path"])
summary = json.loads(summary_path.read_text(encoding="utf-8"))
summary["aggregate_by_mechanism"]
```

## Interpretation

Compare additive controls against XOR and multiplicative-gate mechanisms. PEID's expected advantage is concentrated where the true mechanism is a source-set hyperedge rather than a reducible pairwise edge.
